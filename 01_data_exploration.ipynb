# Product Recommendation System

## Phase 1 - Exploratory Data Analysis

### Objective

Explore the Online Retail dataset to understand product information,
customer purchasing behaviour, transaction patterns and data quality
before developing the recommendation models.

### Dataset

- Source: UCI Machine Learning Repository
- Dataset: Online Retail
- Records: 541,909
- Features: 8

```python
import urllib.request
import zipfile
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://archive.ics.uci.edu/static/public/352/online+retail.zip"

zip_path = "/content/online_retail.zip"

urllib.request.urlretrieve(url, zip_path)

print("Dataset downloaded successfully!")
```

```python
extract_path = "/content/data"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")
```

```python
df = pd.read_excel("/content/data/Online Retail.xlsx")

df.head()
```
